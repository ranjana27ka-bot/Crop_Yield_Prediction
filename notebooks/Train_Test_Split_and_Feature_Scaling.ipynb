Objective
Separate features and target variable.
Split the dataset into training and testing sets.
Scale numerical features using StandardScaler and MinMaxScaler.
Save the scaler for deployment.

Import Required Libraries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler

import joblib
```

Load the Dataset

```python
df = pd.read_csv("../data/crop_yield_selected_features.csv")

df.head()
```

Check Dataset

```python
df.shape
df.info()
```

Separate Features and Target

```python
X = df.drop("Yield", axis=1)

y = df["Yield"]
```

Display Shapes

```python
print("Features Shape :", X.shape)

print("Target Shape :", y.shape)
```

Train-Test Split


```python
#Split the dataset into 80% training and 20% testing.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

Check Shapes

```python

print("Training Features :", X_train.shape)

print("Testing Features :", X_test.shape)

print("Training Target :", y_train.shape)

print("Testing Target :", y_test.shape)
```

Standard Scaling

```python
#Create the scaler.

scaler = StandardScaler()
```

```python
#Fit only on training data.

X_train_scaled = scaler.fit_transform(X_train)
```

```python
#Transform testing data.

X_test_scaled = scaler.transform(X_test)
```

```python
# Convert Back to DataFrame 
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns
)
```

```python
 #View Scaled Data
X_train_scaled.head()
```

```python
#Verify Scaling
print(X_train_scaled.describe().round(2))
```

Min-Max Scaling

```python
minmax = MinMaxScaler()

X_train_minmax = minmax.fit_transform(X_train)

X_test_minmax = minmax.transform(X_test)

#Convert to DataFrame.

X_train_minmax = pd.DataFrame(
    X_train_minmax,
    columns=X_train.columns
)

X_test_minmax = pd.DataFrame(
    X_test_minmax,
    columns=X_test.columns
)
```

Compare StandardScaler and MinMaxScaler

```python
print("Standard Scaled Data")

print(X_train_scaled.head())

print("MinMax Scaled Data")

print(X_train_minmax.head())
```

Save the StandardScaler

```python
joblib.dump(scaler, "standard_scaler.pkl")

print("StandardScaler saved successfully.")
```

Save the MinMaxScaler

```python
joblib.dump(minmax, "minmax_scaler.pkl")

print("MinMaxScaler saved successfully.")
```

Save Training and Testing Data 

```python
from pathlib import Path

# Current notebook location
print(Path.cwd())

# If notebook is in notebooks/
project_root = Path.cwd().parent

data_folder = project_root / "data"
data_folder.mkdir(exist_ok=True)  # Creates the folder if it doesn't exist


X_train_scaled.to_csv(data_folder / "X_train_scaled.csv", index=False)

X_test_scaled.to_csv(data_folder / "X_test_scaled.csv", index=False)

y_train.to_csv(data_folder / "y_train.csv", index=False)

y_test.to_csv(data_folder / "y_test.csv", index=False)
```
